```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
L, width, thickness = 0.10, 0.025, 0.002
k, T_b, T_inf, h = 205.0, 100.0, 25.0, 25.0
x = np.linspace(0.0, L, 101)
A_c = width * thickness
P = 2 * (width + thickness)
m = np.sqrt(h * P / (k * A_c))
T = T_inf + (T_b - T_inf) * np.cosh(m * (L - x)) / np.cosh(m * L)
print('Array shape:', x.shape)
print(f'm = {m:.3f} 1/m; tip temperature = {T[-1]:.2f} deg C')
```

```python
fig, ax = plt.subplots(figsize=(7.2, 4.5), constrained_layout=True)
ax.plot(x * 1000, T, linewidth=2.4, label=f'h = {h:g} W/(m² K)')
ax.set(xlabel='Distance from base, x (mm)', ylabel='Temperature, T (°C)',
       title='Temperature Distribution Along an Aluminum Fin')
ax.legend(frameon=True)
ax.grid(True, alpha=0.25)
plt.show()
```

Each pair `(x[i], T[i])` represents the predicted temperature at one physical location. Equal array shapes allow Matplotlib to map every horizontal coordinate to one vertical coordinate.

```python
h_values = np.array([10.0, 25.0, 50.0, 100.0])
tip_temperatures = []
fig, ax = plt.subplots(figsize=(7.2, 4.5), constrained_layout=True)
colors = plt.cm.viridis(np.linspace(0.12, 0.88, len(h_values)))

for h_case, color in zip(h_values, colors):
    m_case = np.sqrt(h_case * P / (k * A_c))
    T_case = T_inf + (T_b - T_inf) * np.cosh(m_case * (L - x)) / np.cosh(m_case * L)
    ax.plot(x * 1000, T_case, linewidth=2.2, color=color, label=f'h = {h_case:g} W/(m² K)')
    tip_temperatures.append(T_case[-1])

ax.set(xlabel='Distance from base, x (mm)', ylabel='Temperature, T (°C)',
       title='Effect of Convection on Fin Temperature')
ax.legend(title='Convection case', frameon=True)
ax.grid(True, alpha=0.25)
fig.savefig('fin_temperature_sweep.png', dpi=300, bbox_inches='tight')
plt.show()

for h_case, T_tip in zip(h_values, tip_temperatures):
    print(f'h = {h_case:5.1f} W/(m^2 K) -> T_tip = {T_tip:6.2f} deg C')
```

## Interpretation
Increasing the convection coefficient makes the fin cool more rapidly with distance and lowers the tip temperature. Stronger convection transfers more energy from the fin surface to the surroundings. Under the stated assumptions, the $h=100$ W/(m²·K) case provides the strongest convective heat removal, although a full design decision should also consider how the airflow needed to create that value affects fan power and system cost.
