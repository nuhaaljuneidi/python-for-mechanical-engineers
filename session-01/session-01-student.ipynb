## 3. Internal flow: Reynolds number and decisions

For flow in a circular pipe, $Re = \rho VD/\mu$. Classify the flow as laminar ($Re<2300$), transitional ($2300\le Re<4000$), or turbulent ($Re\ge4000$).

```python
rho_kg_m3 = 997.0       # water near room temperature
velocity_m_s = 0.8
diameter_m = 0.025
mu_Pa_s = 0.00089

Re = rho_kg_m3 * velocity_m_s * diameter_m / mu_Pa_s

if Re < 2300:
    regime = "laminar"
elif Re < 4000:
    regime = "transitional"
else:
    regime = "turbulent"

print(f"Re = {Re:,.0f}; the flow is {regime}.")
```

## 4. Heat transfer: a wall as a resistance network

For a plane wall with convection on both sides,
$R''_{total}=1/h_i+L/k+1/h_o$ and $U=1/R''_{total}$.

Use $h_i=10$ W/(m²·K), $h_o=25$ W/(m²·K), insulation thickness $L=0.08$ m, and $k=0.04$ W/(m·K).

```python
h_inside = 10.0
h_outside = 25.0
thickness_m = 0.08
k_W_mK = 0.04

R_conv_inside = 1 / h_inside
R_conduction = thickness_m / k_W_mK
R_conv_outside = 1 / h_outside
R_total = R_conv_inside + R_conduction + R_conv_outside
U_W_m2K = 1 / R_total

print(f"Overall U-value: {U_W_m2K:.3f} W/(m²·K)")
print(f"Insulation fraction of resistance: {100*R_conduction/R_total:.1f}%")
```
